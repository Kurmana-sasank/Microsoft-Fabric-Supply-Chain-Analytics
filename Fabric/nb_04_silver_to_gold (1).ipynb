```python
from pyspark.sql.functions import *

fact_orders = (
    silver_df
    .groupBy("order_id")
    .agg(
        first("customer_id").alias("customer_id"),
        first("order_date").alias("order_date"),
        first("market").alias("market"),
        first("order_region").alias("order_region"),

        sum("sales").alias("total_sales"),
        sum("benefit_per_order").alias("total_profit"),

        avg("delivery_delay_days").alias("avg_delivery_delay_days"),

        max(
            col("is_late_delivery").cast("int")
        ).alias("is_late_delivery"),

        max(
            col("is_cancelled").cast("int")
        ).alias("is_cancelled"),

        sum("order_item_quantity").alias("total_quantity")
    )
    .withColumn(
        "order_date_only",
        to_date(col("order_date"))
    )
)
```

```python
(
    fact_orders.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("fact_orders")
)
```

```python
from pyspark.sql.functions import *

silver_df = spark.table("silver_supply_chain")

silver_df = silver_df.withColumn(
    "profit_margin_pct",
    when(
        col("sales") != 0,
        round((col("benefit_per_order") / col("sales")) * 100, 2)
    ).otherwise(0)
)

(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("silver_supply_chain")
)

print("profit_margin_pct added successfully")
```

```python
from pyspark.sql.functions import *

silver_df = spark.table("silver_supply_chain")

customer_orders = (
    silver_df
    .groupBy("customer_id")
    .agg(
        countDistinct("order_id").alias("total_orders")
    )
)

fact_customer_behavior = (
    silver_df
    .groupBy("customer_id")
    .agg(
        countDistinct("order_id").alias("total_orders"),
        sum("sales").alias("customer_lifetime_value"),
        sum("order_item_quantity").alias("total_quantity"),
        avg("sales").alias("avg_order_value")
    )
    .withColumn(
        "is_repeat_customer",
        when(col("total_orders") > 1, 1).otherwise(0)
    )
)

(
    fact_customer_behavior.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema","true")
    .saveAsTable("fact_customer_behavior")
)

print(
    "fact_customer_behavior:",
    spark.table("fact_customer_behavior").count()
)
```
